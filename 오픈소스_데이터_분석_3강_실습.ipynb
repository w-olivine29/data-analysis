## 3-1 언패킹

### 시퀀스 언패킹

```python
coordinates = (10, 20)

# 튜플 언패킹

x,y = coordinates
print(f'x좌표: {x}, y좌표: {y}')

rgb = [255, 128, 0]
# 리스트 언패킹

red,green,blue = rgb
print(f'빨강: {red}, 초록: {green}, 파랑: {blue}')

word = 'ABC'
# 문자열 언패킹

first, second, third = word
print(f'첫 번째 문자: {first}, 두 번째 문자: {second}, 세 번째 문자: {third}')
```

```python
def get_dataset_stats(values):
  return min(values), max(values), sum(values) / len(values)  #튜플로 반환

data = [12, 8, 21, 17, 5, 32, 14]

# 함수값 return 언패킹
minimum, maximum, average = get_dataset_stats(data)  #반환된 튜플을 언패킹해서 받기

print(f'최솟값: {minimum}, 최댓값: {maximum}, 평균: {average:.2f}')
```

```python
student_scores = [
    ('김철수', 85, 92, 78),
    ('이영희', 92, 88, 95),
    ('박지민', 75, 83, 90)
]

# 반복문에서 언패킹 사용
for name,database, python, cloud in student_scores :
  average = (database + python + cloud) / 3
  print(f'{name}의 평균 점수: {average:.1f}')
```

## 3-3 함수형 프로그래밍

### map

```python
employees = {
    '이지혜': 3000000,
    '구민준': 5000000,
    '방서연': 4100000
}

# 함수 정의: salary * 1.1
adjust_salary = lambda salary : salary * 1.1

# 급여 인상
updated_salaries = {name: adjust_salary(salary) for name, salary in employees.items()}

print(updated_salaries)
```

```python
temperature_celsius = [25.6, 27.8, 30.5, 22.3, 28.9, 31.2, 24.7]

# 함수 정의: 섭씨 -> 화씨
cel_to_fah = lambda temp : (temp * 9/5) + 32

# 온도 변환

#temperature_fahrenheit = {f"{cel}℃" : f"{cel_to_fah(cel)}Ｆ" for cel in temperature_celsius}

# temperature_fahrenheit = list(map(cel_to_fah,temperature_celsius))
temperature_fahrenheit = map(cel_to_fah,temperature_celsius)

print('섭씨 온도 데이터:', temperature_celsius)
print('화씨 온도로 변환:', temperature_fahrenheit)
```

```python
words = ['python', 'data', 'analysis']

# 함수 정의 및 대문자화
uppercase_words = list(map(lambda word : word.upper(), words))

print(uppercase_words)
```

```python
employees = [
    {'name': '김지원', 'salary': 3700000},
    {'name': '박민준', 'salary': 7000000},
    {'name': '이서연', 'salary': 4400000},
    {'name': '정현우', 'salary': 3500000}
]

# salary * 1.1
adjust_salary = lambda employee : {"name":employee["name"], "salary": employee["salary"] * 1.1}
updated_employees = list(map(adjust_salary,employees))

print('\n급여 인상 후:')

for emp in updated_employees:
    print(f'{emp['name']}: {int(emp['salary']):,}원')
```

### filter

```python
numbers = [10, 15, 20, 25, 30]


# 짝수 필터링
even_numbers = list(filter(lambda num : num % 2 == 0, numbers))

print(even_numbers)
```

```python
osda_students = [
    {'name': '김지원', 'score': 85},
    {'name': '이민준', 'score': 65},
    {'name': '박서연', 'score': 90},
    {'name': '정현우', 'score': 55},
    {'name': '최예은', 'score': 78}
]

# 70점 이상 학생 목록 생성
passed_students = list(filter(lambda student : student['score'] >= 70, osda_students))

for student in passed_students:
    print(f'이름: {student["name"]}, 점수: {student["score"]}')
```

### **reduce** & 삼항연산자
reduce: 앞에서부터 2개씩 짝지어 연산한 뒤, 그 결과를 다음 요소와 계속 누적해 단 하나의 최종값을 만드는 함수



```python
from functools import reduce

numbers = [3, 7, 2, 9, 5]

# 최대값 생성
max_value = reduce(lambda num1, num2 : num1 if num1 > num2 else num2, numbers)

max_value = reduce(max, numbers)

print(max_value)
```

- 1단계: 첫 번째(3)와 두 번째(7) 비교 $\rightarrow$ 결과: 7
- 2단계: 앞선 결과(7)와 세 번째(2) 비교 $\rightarrow$ 결과: 7
- 3단계: 앞선 결과(7)와 네 번째(9) 비교 $\rightarrow$ 결과: 9
- 4단계: 앞선 결과(9)와 마지막(5) 비교 $\rightarrow$ 결과: 9
