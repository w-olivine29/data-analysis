### 6-1 데이터 훑어보기

```python
import pandas as pd
import numpy as np

data = {
    'name': ['김민수', '이지영', '박준호', '최서연', '정도윤'],
    'age': [25, 30, 28, 22, 35],
    'city': ['서울', '부산', '인천', '서울', '대전'],
    'score': [90, 85, 95, 80, np.nan]
}
df = pd.DataFrame(data)

# 데이터 측정
df.describe() # 기본설정은 수치형 데이터만 나옴


# 모든 데이터 타입 측정
df.describe(include= "all") # 범주형 데이터도 포함하기 위해 설정

# 수치형 데이터 측정
df.describe(include= [np.number])

# 범주형 데이터 측정
df.describe(include= ["object"])

# 특정 열 분석
df["score"].describe()
```

### 6-2 DataFrame 열 선택 및 **조작**

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}
df = pd.DataFrame(data)

# 특정 열 선택
selected_col = df[["이름", "학과", "학점"]]
print(selected_col, end="\n\n")

# DataFrame 열순서 변경 (원하는 순서로 나열 가능)
reordered_cols = df[["학과", "학년", "이름", "학점"]]
print(reordered_cols, end="\n\n")

# DataFrame 열이름 변경
renamed_cols = df.rename(columns={"이름": "학생명", "학점": "평점"})
print(renamed_cols, end="\n\n")
```

### 6-3 DataFrame 행 선택

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"],
}

# ==============================================================================
# 1. 기본 생성 및 인덱스(인덱스 라벨) 설정
# ==============================================================================

# [1-1] 기본 DataFrame 생성
# > 0, 1, 2... 정수형 인덱스가 자동 부여됨
print(f"\n{'=' * 50}\n[1-1] 기본 DataFrame 형태\n")
df = pd.DataFrame(data)
print(df)


# [1-2] 이름 컬럼을 인덱스로 변경
print(f"\n{'-' * 50}\n[1-2] '이름' 컬럼을 행 인덱스로 지정: df.set_index('이름')\n")

df_indexed = df.set_index("이름") # > 기본 숫자 라벨(0, 1, 2...) 대신 '이름' 컬럼의 데이터들을 행의 라벨(Label)로 변경 (loc 접근 시 사용)
print(df_indexed)


# ==============================================================================
# 2. 라벨 기반 행 선택 (loc)
# ==============================================================================

# [2-1] 범위 슬라이싱
# > 주의: loc 슬라이싱은 기본 숫자 인덱스(0, 1, 2...)여도 '끝 라벨'을 반드시 포함 (iloc는 미포함)

print(f"\n{'=' * 50}\n2. 인덱스 기반 행 선택: df.loc\n\n[실행] df.loc['이영희':'정소희'] (이영희부터 정소희까지 포함)\n")
print(df_indexed.loc["이영희":"정소희"]) # > 이름 컬럼을 인덱스로 설정했기에 문자열 이름값으로 슬라이싱 가능


# ==============================================================================
# 3. 위치 기반 선택 (iloc: index location, integer location)
# ==============================================================================

# [3-1] 단일 행 선택 (Series 반환)
# > 인덱스 라벨 명칭과 무관하게 0부터 시작하는 실제 위치(정수 오프셋)로 추출

print(f"\n{'=' * 50}\n3. 위치 기반 행 선택: df.iloc (정수 인덱스)\n\n[실행] df.iloc[1] (1개 행 추출 -> Series 반환)\n")
print(df_indexed.iloc[1])


# [3-2] 여러 행 선택 (DataFrame 반환)

print(f"\n[실행] df.iloc[[1, 3]] (1번, 3번 행 추출 -> DataFrame 반환)\n")
print(df_indexed.iloc[[1, 3]]) # > 리스트 형태로 정수 번호를 전달하여 복수 행 추출



# ==============================================================================
# 4. 조건 기반 선택
# ==============================================================================

print(f"\n{'=' * 50}\n4. 조건 기반 선택 \n")


# 첫번째와 세번째만 가져옴
print("\n[실행] df.loc[[True, False, True, False, False]]\n")

print(df.loc[[True, False, True, False, False]]) # 들어가는 인자의 수는 레코드 개수와 동일해야함


# 4.0 이상인지 여부(True/False)로 이루어진 시리즈 생성
print("\n[실행] df[\"학점\"] >= 4.0\n")

high_score = df["학점"] >= 4.0  # 단일 True/ False 값이 아닌 여러 값이 나옴
print(high_score)


# loc에 조건 넣기
print("\n[실행] df.loc[df[\"학점\"]]\n")
print(df.loc[high_score])

# df[] 안에 조건 넣기
df[df["학년"] == 2]

df[(df["학년"] == 2) & (df["학점"]>= 3.7)]
```

### 6-4 DataFrame 정렬

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}

#DataFrame 인덱스 지정
df = pd.DataFrame(data).set_index("이름")
print("[이름을 인덱스로 지정한 결과]\n", df, end="\n\n")


#DataFrame 단일 열 오름차순, 내림차순 정렬
df.sort_values(by="학점") # 오름차순 (기본 설정)
df.sort_values(by="학점", ascending= False) # ascending= False -> 내림차순

#DataFrame 복수 열 오름차순, 내림차순 정렬
df.sort_values(by=["학년", "학점"], ascending= [True, False]) # 학년 오름차순, 학점 내림차순으로 정렬


#DataFrame 인덱스 기준 오른차순, 내림차순 정렬
index_sorted = df.sort_index(ascending= False)
print("[인덱스 기준 정렬 - 이름 순 내림차순 정렬]\n",index_sorted)
```
